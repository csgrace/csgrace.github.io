### Snake: Simulate a snake game and print the game results.

You are given a map that ‘x’ represents a rock, ‘-’represents a space, ‘#’represents the body of snake. ‘@’represent the head of the snake and a sequence of actions that ‘0,1,2,3’represent to move to up/down/left/right correspondingly for one step.
A greedy snake starts in the map state and moves one step per unit of time according to the sequence of actions until all actions complete or fail. It will fail when the head and the stone overlap, the head goes beyond the boundary, or the head overlaps the body. 

#### Input
A matrix with type char (the map). 
A sequence with type int (the motions). 

#### Output
the the result of the game:
If it failed, output the running time of the game.
It it didn’t fail, output the final position of the head (in the form “%d, %d”).

```python
"""
Example:
input:
map:
---------
------x--
-x-------
---@-----
---##----
------x--
--x----x-
-x-------
---------
action:
0 0 3 3 0 3 3 1 1 1 1 1 3 1 1 2 2 2 2 2

output:
7 3
"""
```

```python
def simulate_snake(game_map, actions):

    rows = len(game_map)
    cols = len(game_map[0])

    head_pos = None
    for i in range(rows):
        for j in range(cols):
            if game_map[i][j] == '@':
                head_pos = (i, j)
                break
        if head_pos is not None:
            break
    if head_pos is None:
        return "0"

    snake_queue = [head_pos]
    visited = set([head_pos])
    current = head_pos
    found = True
    directions = [(0, 1), (0, -1), (1, 0), (-1, 0)]  # right, left, down, up
    while found:
        found = False
        i, j = current
        for di, dj in directions:
            ni, nj = i + di, j + dj
            if 0 <= ni < rows and 0 <= nj < cols:
                if game_map[ni][nj] == '#' and (ni, nj) not in visited:
                    visited.add((ni, nj))
                    snake_queue.append((ni, nj))
                    current = (ni, nj)
                    found = True
                    break
    move_dirs = [(-1, 0), (1, 0), (0, -1), (0, 1)]
    for step, act in enumerate(actions):
        di, dj = move_dirs[act]
        head_i, head_j = snake_queue[0]
        new_i = head_i + di
        new_j = head_j + dj
        new_head = (new_i, new_j)

        if new_i < 0 or new_i >= rows or new_j < 0 or new_j >= cols:
            return str(step + 1)

        if game_map[new_i][new_j] == 'x':
            return str(step + 1)

        if len(snake_queue) > 1:
            if new_head in snake_queue[1:-1]:
                return str(step + 1)

        # Move the snake: add new head and remove the tail
        snake_queue.insert(0, new_head)
        snake_queue.pop()

    # All actions completed without failure
    return f"{snake_queue[0][0]}, {snake_queue[0][1]}"

```

```python
def run_test_case(test_case):
    base_path = 'D:/E/SUSTECH/grade3/3_up/CS/AI/lab/lab1/Practice1/test_cases/problem3'
    with open(f'{base_path}/{test_case}-map.txt', 'r') as f:
        game_map = [list(line.strip()) for line in f]
    print(game_map)
    with open(f'{base_path}/{test_case}-actions.txt', 'r') as f:
        actions = list(map(int, f.read().split()))
    print(actions)
    print(simulate_snake(game_map, actions))

for test_case in range(1, 5):
    run_test_case(test_case)
```
